```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import matplotlib.pyplot as plt
```

```python
# X = study hours
X = np.array([1, 2, 3, 4, 5, 6, 7, 8], dtype=float)

# y = marks scored
y = np.array([35, 40, 50, 55, 65, 70, 80, 85], dtype=float)

print("Study hours:", X)
print("Marks:", y)
```

```python
model = Sequential([
    Dense(1, input_shape=[1])
])
```

```python
model.compile(
    optimizer='sgd',
    loss='mean_squared_error'
)
```

```python
history = model.fit(X, y, epochs=500, verbose=0)
```

```python
hours = 8.5
prediction = model.predict(np.array([hours]))

print(f"If student studies {hours} hours, predicted marks = {prediction[0][0]:.2f}")
```

```python
plt.plot(history.history['loss'])
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Loss Decreasing During Training")
plt.show()
```
