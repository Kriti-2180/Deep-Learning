```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([1, 2, 3, 4, 5, 6], dtype=float)

y = np.array([0, 0, 0, 1, 1, 1], dtype=float)
```

```python
weight = 0.5
bias = 0.0

learning_rate = 0.1
epochs = 1000
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))


loss_history = []
```

```python
for epoch in range(epochs):

    z = X * weight + bias

    prediction = sigmoid(z)

    loss = np.mean((prediction - y) ** 2)

    loss_history.append(loss)

    dLoss_dPrediction = (
        2 * (prediction - y) / len(y)
    )


    dPrediction_dZ = (
        prediction * (1 - prediction)
    )


    dZ_dWeight = X


    dLoss_dWeight = np.sum(
        dLoss_dPrediction
        * dPrediction_dZ
        * dZ_dWeight
    )


    dLoss_dBias = np.sum(
        dLoss_dPrediction
        * dPrediction_dZ
    )

    weight = (
        weight
        - learning_rate * dLoss_dWeight
    )

    bias = (
        bias
        - learning_rate * dLoss_dBias
    )

    if epoch % 100 == 0:
        print(
            f"Epoch: {epoch}, "
            f"Loss: {loss:.4f}, "
            f"Weight: {weight:.4f}, "
            f"Bias: {bias:.4f}"
        )
```

```python
print("\nFinal Weight:", weight)
print("Final Bias:", bias)
```

```python
print("\nNumber of stored losses:", len(loss_history))
```

```python
plt.figure(figsize=(8, 5))

plt.plot(loss_history)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Loss Decreasing During Backpropagation")

plt.show()
```

```python
study_hours = 5

z = study_hours * weight + bias

prediction = sigmoid(z)

print("Prediction Probability:", prediction)

if prediction >= 0.5:
    print("Prediction: PASS")
else:
    print("Prediction: FAIL")
```
