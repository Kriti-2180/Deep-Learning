```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import matplotlib.pyplot as plt
```

```python
X = np.array([
    [1, 4],
    [2, 5],
    [3, 4],
    [4, 6],
    [5, 7],
    [6, 7],
    [7, 8],
    [8, 8]
], dtype=float)
```

```python
y = np.array([
    0,
    0,
    0,
    0,
    1,
    1,
    1,
    1
], dtype=float)
```

```python
model = Sequential([
    Dense(4, input_shape=(2,), activation='sigmoid'),
    Dense(1, activation='sigmoid')
])
```

```python
model.compile(
    optimizer='sgd',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(X, y, epochs=1000, verbose=2)
```

```python
student = np.array([[6, 7]])
prediction = model.predict(student)

print("Prediction value:", prediction[0][0])

if prediction[0][0] >= 0.5:
    print("Result: Pass")
else:
    print("Result: Fail")
```

```python
plt.plot(history.history['loss'])
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Loss Decreases Because of Backpropagation")
plt.show()
```

```python
plt.plot(history.history['accuracy'])
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.title("Accuracy Increases During Training")
plt.show()
```
