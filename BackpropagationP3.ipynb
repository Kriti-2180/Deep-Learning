```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([1, 2, 3, 4], dtype=float)

y = np.array([2, 4, 6, 8], dtype=float)
```

```python
def mse(y_true, y_pred):
    return np.mean(
        (y_true - y_pred) ** 2
    )
```

```python
np.random.seed(42)

random_weight = 0.5

random_loss_history = []

random_weight_history = []
```

```python
for epoch in range(100):

    prediction = X * random_weight

    current_loss = mse(y, prediction)

    random_loss_history.append(current_loss)

    random_weight_history.append(random_weight)


    candidate_weight = (
        random_weight
        + np.random.uniform(-0.5, 0.5)
    )


    candidate_prediction = (
        X * candidate_weight
    )


    candidate_loss = mse(
        y,
        candidate_prediction
    )


    if candidate_loss < current_loss:
        random_weight = candidate_weight
```

```python
weight = 0.5

learning_rate = 0.01

epochs = 100

backprop_loss_history = []

backprop_weight_history = []
```

```python
for epoch in range(epochs):

    prediction = X * weight

    loss = mse(y, prediction)

    backprop_loss_history.append(loss)

    backprop_weight_history.append(weight)


    gradient = np.mean(
        2 * (prediction - y) * X
    )



    weight = (
        weight
        - learning_rate * gradient
    )


    if epoch % 10 == 0:

        print(
            f"Epoch: {epoch}, "
            f"Loss: {loss:.6f}, "
            f"Gradient: {gradient:.6f}, "
            f"Weight: {weight:.6f}"
        )
```

```python
print("\n-----------------------")
print("FINAL RESULTS")
print("-----------------------")

print(
    "Random Search Weight:",
    random_weight
)

print(
    "Backprop Weight:",
    weight
)

print(
    "Random Search Loss:",
    random_loss_history[-1]
)

print(
    "Backprop Loss:",
    backprop_loss_history[-1]
)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    random_loss_history,
    label="Random Search"
)

plt.plot(
    backprop_loss_history,
    label="Backpropagation"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.title(
    "Random Guessing vs Backpropagation"
)

plt.legend()

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    random_weight_history,
    label="Random Search Weight"
)

plt.plot(
    backprop_weight_history,
    label="Backprop Weight"
)

plt.axhline(
    y=2.0,
    linestyle="--",
    label="Correct Weight = 2"
)

plt.xlabel("Epoch")
plt.ylabel("Weight")

plt.title(
    "How Weight Moves Toward Correct Value"
)

plt.legend()

plt.show()
```
