```python
import os
import cv2
import matplotlib.pyplot as plt

# dataset path
path = "/content/asl_alphabet_train/asl_alphabet_train"
class_name = "A"

# pick one image automatically
img_name = os.listdir(os.path.join(path, class_name))[0]
img_path = os.path.join(path, class_name, img_name)

# read image
img = cv2.imread(img_path)

# preprocessing
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
blur = cv2.GaussianBlur(gray, (5,5), 0)
_, thresh = cv2.threshold(blur, 120, 255, cv2.THRESH_BINARY)

# display all
plt.figure(figsize=(10,5))

plt.subplot(1,4,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title(f"Original ({class_name})")
plt.axis('off')

plt.subplot(1,4,2)
plt.imshow(gray, cmap='gray')
plt.title("Gray")
plt.axis('off')

plt.subplot(1,4,3)
plt.imshow(blur, cmap='gray')
plt.title("Blur")
plt.axis('off')

plt.subplot(1,4,4)
plt.imshow(thresh, cmap='gray')
plt.title("Threshold")
plt.axis('off')

plt.tight_layout()
plt.show()
```

#2. Using function

```python
def show_preprocessing(path, class_name):
    import os, cv2, matplotlib.pyplot as plt

    img_name = os.listdir(os.path.join(path, class_name))[0]
    img_path = os.path.join(path, class_name, img_name)

    img = cv2.imread(img_path)

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    blur = cv2.GaussianBlur(gray, (5,5), 9)
    _, thresh = cv2.threshold(blur, 120, 255, cv2.THRESH_BINARY)

    plt.figure(figsize=(10,5))

    plt.subplot(1,4,1)
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title(f"{class_name}")
    plt.axis('off')

    plt.subplot(1,4,2)
    plt.imshow(gray, cmap='gray')
    plt.title("Gray")
    plt.axis('off')

    plt.subplot(1,4,3)
    plt.imshow(blur, cmap='gray')
    plt.title("Blur")
    plt.axis('off')

    plt.subplot(1,4,4)
    plt.imshow(thresh, cmap='gray')
    plt.title("Threshold")
    plt.axis('off')

    plt.show()
```

```python
show_preprocessing(path, "A")
show_preprocessing(path, "B")
```

```python
import os
import cv2
import matplotlib.pyplot as plt

path = "/content/asl_alphabet_train/asl_alphabet_train"

# 👇 CHOOSE YOUR LETTER HERE
class_name = "D"   # change to B, C, D...

# pick image from that class
img_name = os.listdir(os.path.join(path, class_name))[0]
img_path = os.path.join(path, class_name, img_name)

# read image
img = cv2.imread(img_path)
```

```python
import matplotlib.pyplot as plt

resized = cv2.resize(img, (128,128), interpolation=cv2.INTER_CUBIC)

plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
plt.title("Resized (128x128)")
plt.axis('off')

plt.show()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Apply Histogram Equalization
equalized = cv2.equalizeHist(gray)

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(gray, cmap='gray')
plt.title("Original Gray")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(equalized, cmap='gray')
plt.title("Enhanced Contrast")
plt.axis('off')

plt.show()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Create CLAHE object
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))

# Apply CLAHE
enhanced = clahe.apply(gray)
```

```python
plt.imshow(enhanced, cmap='gray')
plt.title("CLAHE Output (Best)")
plt.axis('off')
```
