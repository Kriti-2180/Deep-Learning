```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def loss_function(x):
    return x**2


x = np.linspace(-10,10,100)

y = loss_function(x)


plt.figure(figsize=(8,5))

plt.plot(x,y)

plt.xlabel("x value")
plt.ylabel("Loss")

plt.title("Loss Function")

plt.grid()

plt.show()
```

```python
def gradient(x):
    return 2*x
```

```python
def gradient_descent(
    start_point,
    learning_rate,
    iterations
):

    x = start_point

    history=[]


    for i in range(iterations):

        grad = gradient(x)

        x = x - learning_rate * grad

        history.append(x)


    return history
```

```python
history = gradient_descent(
    start_point=10,
    learning_rate=0.1,
    iterations=50
)


print("Final value:",history[-1])
```

```python
plt.figure(figsize=(8,5))


plt.plot(
    history,
    marker='o'
)


plt.xlabel("Iteration")

plt.ylabel("x value")

plt.title(
"Gradient Descent Weight Updates"
)


plt.grid()

plt.show()
```

```python
x_values = np.linspace(-10,10,100)

y_values = x_values**2


plt.figure(figsize=(8,5))


plt.plot(
    x_values,
    y_values
)


history_array=np.array(history)


plt.scatter(
    history_array,
    history_array**2,
    color="red"
)


plt.xlabel("x")

plt.ylabel("Loss")


plt.title(
"Gradient Descent Finding Minimum"
)


plt.grid()

plt.show()
```

```python
history_small_lr = gradient_descent(
    start_point=10,
    learning_rate=0.01,
    iterations=100
)


plt.plot(
    history_small_lr,
    label="LR=0.01"
)


plt.legend()

plt.title(
"Small Learning Rate"
)

plt.grid()

plt.show()
```

```python
history_large_lr = gradient_descent(
    start_point=10,
    learning_rate=0.9,
    iterations=20
)


plt.plot(
    history_large_lr,
    label="LR=0.9"
)


plt.legend()

plt.title(
"Large Learning Rate"
)

plt.grid()

plt.show()
```

```python
learning_rates=[
    0.001,
    0.01,
    0.1,
    0.5
]


plt.figure(figsize=(8,5))


for lr in learning_rates:

    history=gradient_descent(
        10,
        lr,
        50
    )


    plt.plot(
        history,
        label=f"LR={lr}"
    )


plt.xlabel("Iterations")

plt.ylabel("x value")

plt.title(
"Effect of Learning Rate"
)


plt.legend()

plt.grid()

plt.show()
```

```python
from sklearn.datasets import make_regression


X,y = make_regression(
    n_samples=1000,
    n_features=1,
    noise=20,
    random_state=42
)


plt.scatter(
    X,
    y
)

plt.xlabel("X")

plt.ylabel("Y")

plt.title(
"Regression Dataset"
)

plt.show()
```

```python
import tensorflow as tf

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import Dense



model = Sequential([

    Dense(
        16,
        activation="relu",
        input_shape=(1,)
    ),

    Dense(1)

])


model.summary()
```

```python
model_batch = Sequential([

    Dense(
        16,
        activation="relu",
        input_shape=(1,)
    ),

    Dense(1)

])


model_batch.compile(

optimizer=tf.keras.optimizers.SGD(
    learning_rate=0.01
),

loss="mse"

)


history_batch=model_batch.fit(

X,
y,

epochs=50,

batch_size=1000,

verbose=1

)
```

```python
model_sgd = Sequential([


    Dense(
        16,
        activation="relu",
        input_shape=(1,)
    ),

    Dense(1)

])



model_sgd.compile(

optimizer=tf.keras.optimizers.SGD(
    learning_rate=0.01
),

loss="mse"

)



history_sgd=model_sgd.fit(

X,
y,

epochs=50,

batch_size=1,

verbose=1

)
```

```python
model_mini = Sequential([


    Dense(
        16,
        activation="relu",
        input_shape=(1,)
    ),

    Dense(1)

])



model_mini.compile(

optimizer=tf.keras.optimizers.SGD(
    learning_rate=0.01
),

loss="mse"

)



history_mini=model_mini.fit(

X,
y,

epochs=50,

batch_size=32,

verbose=1

)
```

```python
plt.figure(figsize=(8,5))


plt.plot(
    history_batch.history["loss"],
    label="Batch GD"
)


plt.plot(
    history_sgd.history["loss"],
    label="SGD"
)


plt.plot(
    history_mini.history["loss"],
    label="Mini Batch GD"
)



plt.xlabel("Epoch")

plt.ylabel("Loss")


plt.title(
"Comparison of Gradient Descent Methods"
)


plt.legend()

plt.grid()

plt.show()
```

```python
test_value=np.array([[5]])


prediction=model_mini.predict(
    test_value
)


print(
"Prediction:",
prediction[0][0]
)
```
